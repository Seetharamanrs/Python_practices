```python
def multiply(a, b):
    result = a * b
    return result

answer = multiply(4, 3)
print(answer)

answer = multiply(answer, 2)
print(answer)
```

```python
def add(a, b):
    return a + b

total = add(5, 3) * 2
print(total)    
```

```python
def subtract(a, b):
    return a - b

result = subtract(15, 6)
print(result)

result = subtract(result, 4)
print(result)
```
