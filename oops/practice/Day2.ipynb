```python
# Condtions 
mark = 85

if mark >= 80:
    print("Distinction")
elif mark >= 60:
    print("First class")
elif mark >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
marks=[90,75,45,30,60,40]
for mark in marks:
    if mark>=80:
        print(mark,"Distinction")
    elif mark>=60:
        print(mark,"First class")
    elif mark>=40:
        print(mark,"Pass")
    else:
        print(mark,"Fail")
```
