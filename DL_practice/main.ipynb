# DEEP LEARNING

## perceptron

```python
x_input=[0.2,0.2,0.2,0.5]
w_weight=[0.1,0.2,0.3]
thersold=0.5
def step(weighted_sum):
    if weighted_sum>thersold:
        return 1
    else:
        return 0

def perceptron():
    weighted_sum=0
    for x,w in zip(x_input,w_weight):
        weighted_sum+=x*w
        print(weighted_sum)
    return step(weighted_sum)

output=perceptron()
print(output)
```

```python
# Build Your First Neuron From Scratch
# Inputs
import numpy as np
x = np.array([2.0, 3.0])

# Weights
w = np.array([0.5, -1.0])

# Bias
b = 1.0
```

```python
## calculating the weight sum 
z=0
for i in range(len(x)):
    z=z+x[i]*w[i]
z=z+b
z
```

```python
def relu(z):
    if z<0:
        return 0
    return z
```

```python
output=relu(z)
print("weight Sum:",z)
print("Neuron output:",output)
```

```python
x=2.0
w=0.5
b=0.0
target=4.0

prediction=x*w+b
print("Prediction:", prediction)
print("Target:", target)
```

```python
# error calcultion
error=target-prediction
error
```

```python
# update the weight
learning_rate=0.1
w=w+learning_rate * error* x
w
```
